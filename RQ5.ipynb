```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
os.makedirs('outputs/figures', exist_ok=True)
os.makedirs('outputs/tables', exist_ok=True)
```

```python
df = pd.read_csv(r'/Users/a/Downloads/spotify_churn_dataset.csv')
```

```python
age_summary = df.groupby('age')['is_churned'].mean().reset_index()

age_summary.to_csv(
    'outputs/tables/rq5_age_churn.csv',
    index=False
)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=age_summary,
            x='age',
            y='is_churned')

plt.title('age vs is_churned Rate')
plt.tight_layout()
plt.savefig('outputs/figures/rq5_age_vs_churn.pdf')
plt.show()
```
