```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
os.makedirs('outputs/figures', exist_ok=True)
os.makedirs('outputs/tables', exist_ok=True)
```

```python
df = pd.read_csv(r'/Users/a/Downloads/spotify_churn_dataset.csv')

summary = pd.crosstab(df['subscription_type'], df['is_churned'])

summary.to_csv('outputs/tables/rq2_premium_vs_churn.csv')
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,
              x='subscription_type',
              hue='is_churned')

plt.title('Premium subscription vs is_churned')
plt.tight_layout()
plt.savefig('outputs/figures/rq2_premium_vs_churn.pdf')
plt.show()
```
